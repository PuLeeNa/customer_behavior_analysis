```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))

# filling null values of RR by each categories median value it belongs. exampls-if clothing RR is null ,
# it will be only filled with median of clothing RR (not whole RR's median).
```

```python
df.isnull().sum()
```

```python
# converting column names into snakecase
df.columns = df.columns.str.lower().str.replace(' ', '_')
```

```python
df.columns
```

```python
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
# create a column age_group
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)
# qcut() divides the data into equal-sized groups based on percentiles, then assigns the labels accordingly.
```

```python
df[['age', 'age_group']].head(10)
```

```python
# create column purchase_frequency_days

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
import subprocess
import sys

def setup_postgresql():
    try:
        # Install PostgreSQL
        subprocess.run(['sudo', 'apt-get', 'update'], check=True)
        subprocess.run(['sudo', 'apt-get', 'install', '-y', 'postgresql', 'postgresql-contrib'], check=True)

        # Start PostgreSQL
        subprocess.run(['sudo', 'service', 'postgresql', 'start'], check=True)

        # Setup user and database
        subprocess.run(['sudo', '-u', 'postgres', 'psql', '-c', "ALTER USER postgres WITH PASSWORD 'pulina123';"], check=True)
        subprocess.run(['sudo', '-u', 'postgres', 'psql', '-c', "CREATE DATABASE customer_behaviour;"], check=True)

        print("PostgreSQL setup complete!")
        return True
    except Exception as e:
        print(f"Setup failed: {e}")
        return False

# Run setup
setup_postgresql()
```

```python
from sqlalchemy import create_engine

# Step 1: Connect to PostgreSQL
username = 'postgres'
password = 'pulina123'
host = 'localhost'
port = '5432'
database = 'customer_behaviour'

engine = create_engine(f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}")

# Step 2: Load DataFrame into PostgreSQL
table_name = 'customer'
df.to_sql(table_name, engine, if_exists="replace", index=False, schema='public')

print(f"Data successfuly loaded into table '{table_name}' in database '{database}'")
```
